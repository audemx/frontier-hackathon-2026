# Configuración e Importaciones

```python
import json
from pathlib import Path
import pandas as pd

# Rutas base
BASE_SAMPLE = Path("../sample")
INPUT_DIR = BASE_SAMPLE / "input"
EXPECTED_DIR = BASE_SAMPLE / "expected_output"

print("Directorio sample encontrado:", BASE_SAMPLE.exists())
```

# Explorando las Tablas

```python
# Cargar clientes.csv con el charset correcto
clientes_path = INPUT_DIR / "tables" / "clientes.csv"
if clientes_path.exists():
  df_clientes = pd.read_csv(clientes_path, encoding="iso-8859-1", sep=";")
  print("--- CLIENTES.CSV ---")
  print("Dimensiones:", df_clientes.shape)
  display(df_clientes.head(3))
else:
  print("Archivo de clientes no encontrado en sample.")
```

# Explorando los Chats de Slack

```python
# Cargar usuarios del chat
users_path = INPUT_DIR / "chats" / "users.json"
if users_path.exists():
  with open(users_path, "r", encoding="utf-8") as f:
    users = json.load(f)
  print(f"Total de usuarios en sample: {len(users)}")
  print("Ejemplo de usuario:", users[0] if users else "N/A")

# Cargar mensajes de un canal y fecha específica del sample
chat_msg_path = INPUT_DIR / "chats" / "soporte-cobros" / "2026-08-03.json"
if chat_msg_path.exists():
  with open(chat_msg_path, "r", encoding="utf-8") as f:
    messages = json.load(f)
  print(f"\nMensajes en el canal de soporte: {len(messages)}")
  print("Ejemplo de mensaje real:")
  print(json.dumps(messages[0], indent=2, ensure_ascii=False))
```

# Explorando Documentos

```python
docs_index_path = INPUT_DIR / "docs" / "index.json"
if docs_index_path.exists():
  with open(docs_index_path, "r", encoding="utf-8") as f:
    docs_index = json.load(f)
  print("Índice de documentos:", docs_index)

# Leer un documento markdown de ejemplo
doc_file_path = INPUT_DIR / "docs" / "memo_cierre_julio.md"
if doc_file_path.exists():
  with open(doc_file_path, "r", encoding="utf-8") as f:
    doc_content = f.read()
  print("\n--- CONTENIDO DEL DOC ---")
  print(doc_content[:500] + "...")
```
